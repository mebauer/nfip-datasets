```python
# import packages
import duckdb
import pandas as pd
```

```python
# reproducibility
%reload_ext watermark
%watermark -v -p duckdb,pandas
```

```python
# count total rows using a glob pattern to match all NYC claims Parquet files
duckdb.sql("""
    SELECT count(*) AS count_rows
    FROM read_parquet('data/claims-*.parquet')
""")
```

```python
# export sample query result to CSV using DuckDB directly
duckdb.sql("""
    COPY (
        SELECT *
        FROM read_parquet('data/claims-*.parquet')
        LIMIT 10
    ) TO 'data/claims-sample.csv' (HEADER, DELIMITER ',')
""")
```

```python
# sanity check
path = 'data/claims-sample.csv'
sample_df = pd.read_csv(path)

# preview sample data
sample_df
```
